```python
import pandas as pd 
import caliperpy
import os
import matplotlib.pyplot as plt
import numpy as np

def read_bin(filename):
    dk = caliperpy.TransCAD.connect()
    out = None
    try:
        out = dk.GetDataFrameFromBin(filename)
    except Exception as e:
        print(e) # type: ignore
    finally:
        dk.Close()
        caliperpy.TransCAD.disconnect()
    return out

SCENARIO_BASE_FOLDER = r'C:\models\Reno_TDM\scenarios'
SCENARIO_CONTROL = 'base_2022'
SCENARIO_CHANGE = 'base_2022_with_resnets'

net_table = read_bin(os.path.join(SCENARIO_BASE_FOLDER, SCENARIO_CONTROL, 'output', 'networks', 'scenario_links.bin'))

# Congestion - is TB-Resnets worse/better than base?

# Validation - is TB-Resnets closer to counts?
```
